```python
# Convert Price from string/object format to numeric format.
# The function removes commas from values such as "1,700"
# so that they can be converted to a number (1700.0).
# This conversion is necessary before performing mathematical
# operations such as Revenue = Quantity Ordered × Price.
```

```python
def obj_to_num(a):
  b=a.split(',')
  a=float(''.join(b))
  return a
```

```python
df['Price']=df['Price'].apply(obj_to_num)
df['Price'].dtypes
```

```python
#Time set to morning afternoon Evening Night
```

```python
df['Time']
```

```python
def change_time_to_range(a):
  b=a.split()
  b0=b[0].split(':')
  if b[1]=='AM':
    if 11>=int(b0[0])>=5 :
      return 'Morning'
    return 'Night'
  else:
    if 12>=int(b0[0])>=5:
      return 'Afternoon'
    return 'Evening'
```

```python
df["Time_range"]=df['Time'].apply(change_time_to_range)
```

```python
df["Time_range"].unique()
```

```python
df.groupby('Time_range')['Quantity Ordered'].sum().sort_values(ascending=False).head(5)
```
